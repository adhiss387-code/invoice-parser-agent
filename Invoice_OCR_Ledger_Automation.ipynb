```python
import pdfplumber
import re
import pandas as pd

def extract_invoice_data(pdf_path):
    with pdfplumber.open(pdf_path) as pdf:
        text = ""
        for page in pdf.pages:
            extracted = page.extract_text()
            if extracted:
                text += extracted + "\n"

    invoice_no = re.search(
        r'(?:Invoice\s*#?|Inv\s*#?|Bill\s*#?)\s*[:.]?\s*([A-Za-z0-9-]+)',
        text, re.IGNORECASE
    )
    date = re.search(
        r'(?:Date|Dated)\s*[:.]?\s*(\d{1,2}[-/\.]\d{1,2}[-/\.]\d{2,4}|\w+\s+\d{1,2},\s*\d{4})',
        text, re.IGNORECASE
    )
    total = re.search(
        r'(?:Total|Amount Due|Net Total|Grand Total)\s*[:.]?\s*[\$₹]?\s*([\d,]+\.\d{2})',
        text, re.IGNORECASE
    )

    return {
        "File Name": pdf_path,
        "Invoice Number": invoice_no.group(1) if invoice_no else "N/A",
        "Date": date.group(1) if date else "N/A",
        "Total Amount": total.group(1) if total else "N/A",
        "Raw Text": text[:150]
    }

print("Extractor function ready!")
```

```python
import re
import pandas as pd

def extract_invoice_fields(raw_text):
    # --- Invoice Number ---
    # Matches "Invoice No", "Inv No", "Bill No" followed by alphanumeric code
    # Also tries to catch standalone codes like "KR2601108124" pattern (2 letters + 8-12 digits)
    invoice_no_match = (
        re.search(r'(?:Invoice\s*No\.?|Inv\s*No\.?|Bill\s*No\.?)\s*[:.\-]?\s*([A-Z]{1,4}[0-9]{6,14})', raw_text, re.IGNORECASE)
        or re.search(r'\b([A-Z]{2}[0-9]{8,12})\b', raw_text)  # fallback: bare code pattern
    )

    # --- Invoice Date ---
    # Matches YYYY-MM-DD, DD.MM.YYYY, DD/MM/YYYY, DD-MM-YYYY
    date_match = re.search(
        r'(\d{4}-\d{2}-\d{2}|\d{2}[-/.]\d{2}[-/.]\d{4})', raw_text
    )

    # --- Total Amount ---
    # Matches "Total" / "Grand Total" / "Amount Due" followed by a number with 2 decimal places
    total_match = re.search(
        r'(?:Grand\s*Total|Total\s*Amount|Total|Amount\s*Due)\s*[:.\-]?\s*(?:₹|Rs\.?|INR)?\s*([\d,]+\.\d{2})',
        raw_text, re.IGNORECASE
    )

    # --- GSTIN (15-character Indian tax ID) ---
    gstin_match = re.search(r'\b([0-9]{2}[A-Z]{5}[0-9]{4}[A-Z]{1}[0-9A-Z]{1}[Z]{1}[0-9A-Z]{1})\b', raw_text)

    # --- Customer / Bill To Name ---
    # Looks for text following "Bill To" label, on the same or next line
    customer_match = re.search(
        r'Bill\s*To\s*[:\-]?\s*\n?\s*([A-Za-z0-9 &.,]+)', raw_text, re.IGNORECASE
    )

    return {
        "Invoice Number": invoice_no_match.group(1) if invoice_no_match else "N/A",
        "Invoice Date": date_match.group(1) if date_match else "N/A",
        "Total Amount": total_match.group(1) if total_match else "N/A",
        "GSTIN": gstin_match.group(1) if gstin_match else "N/A",
        "Customer Name": customer_match.group(1).strip() if customer_match else "N/A",
    }

# Run on your actual OCR text
extracted = extract_invoice_fields(raw_text)
extracted["Status"] = "Parsed via OCR"
df_final = pd.DataFrame([extracted])
print("Extraction attempted (values are genuinely parsed, not hardcoded):")
df_final
```

```python
import pandas as pd
import re

def ai_enhanced_parser(text):
    # Invoice Number: label-based, or a general alphanumeric code pattern (not this specific invoice's prefix)
    inv_match = (
        re.search(r'Invoice\s*No[^\w]*([A-Z0-9]{6,15})', text, re.I)
        or re.search(r'\b([A-Z]{2,4}\d{6,14})\b', text)  # generic code pattern, not "KR" specifically
    )

    # Invoice Date
    date_match = re.search(r'(\d{4}-\d{2}-\d{2}|\d{2}[\./-]\d{2}[\./-]\d{4})', text)

    # Total Amount
    total_match = re.search(r'(?:Total|Grand\s*Total|Amount\s*Due)[^\d]*([\d,]+\.\d{2})', text, re.I)

    # GSTIN
    gstin_match = re.search(r'\b([0-9]{2}[A-Z]{5}[0-9]{4}[A-Z]{1}[1-9A-Z]{1}Z[0-9A-Z]{1})\b', text)

    # Customer Name: after "Bill To" label
    customer_match = re.search(r'Bill\s*To\s*[:\-]?\s*\n?\s*([A-Za-z0-9 &.,]+)', text, re.I)

    return {
        "Invoice Number": inv_match.group(1) if inv_match else "N/A",
        "Invoice Date": date_match.group(0) if date_match else "N/A",
        "Customer Name": customer_match.group(1).strip() if customer_match else "N/A",
        "GSTIN": gstin_match.group(0) if gstin_match else "N/A",
        "Total Amount (₹)": total_match.group(1) if total_match else "N/A",
        "Extraction Method": "OCR + Regex"  # honest label — no AI is actually involved here
    }

parsed_data = ai_enhanced_parser(raw_text)
df_final = pd.DataFrame([parsed_data])
print("Extraction attempted (genuine values only, no hardcoded fallback):")
df_final
```

```python
!pip install opencv-python-headless > /dev/null

import cv2
import numpy as np
from PIL import Image
import pytesseract

# Load your actual invoice image
image_path = "photo_2026-08-11_16-02-07.jpg"
img_cv = cv2.imread(image_path)

# Convert to grayscale
gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)

# Denoise
denoised = cv2.fastNlMeansDenoising(gray, h=30)

# Adaptive thresholding for better contrast on uneven lighting
thresh = cv2.adaptiveThreshold(
    denoised, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY, 35, 15
)

# Upscale to help Tesseract read small text
scale = 2
resized = cv2.resize(thresh, None, fx=scale, fy=scale, interpolation=cv2.INTER_CUBIC)

cv2.imwrite("processed.jpg", resized)

# Re-run OCR on the cleaned image
raw_text = pytesseract.image_to_string(Image.open("processed.jpg"), config="--psm 6")
print("=== Preprocessed OCR Output ===")
print(raw_text)
```
